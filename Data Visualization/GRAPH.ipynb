```python
import matplotlib.pyplot as plt
import numpy as np
import statistics as st
import scipy.stats as ss
```

```python
x=np.arange(-20,20,.01)
y=1/(x**2+1)
plt.figure(figsize=(100,5))
plt.subplot(141)
plt.grid()
plt.plot(x,y,label='y=1/(x^2+1)')
plt.legend()
plt.figure(figsize=(20,5))
plt.subplot(142)
plt.grid()
z=np.sin(x)+x
plt.plot(x,z,label='y=sin(x)+x')
plt.legend()
plt.figure(figsize=(50,5))
plt.subplot(143)
w=x*np.sin(x)
plt.grid()
plt.plot(x,w,label='y=x*sin(x)')
plt.legend()
plt.figure(figsize=(20,5))
plt.subplot(144)
n=ss.norm.pdf(x,st.mean(x),st.stdev(x))
n1=ss.td
plt.grid()
plt.plot(x,n,label='Normal Distribution')
plt.legend()

```

```python
mu, sigma = 0, 0.1
s = np.random.normal(mu,sigma, 1000)
count, bins, ignored = plt.hist(s, 30, density=True)#return the <count> which is the density of the bins, that is their y value, the <bins>which is their lower edgelocation
plt.plot(bins, 1/(sigma * np.sqrt(2 * np.pi)) *np.exp( - (bins - mu)**2 / (2 * sigma**2) ),linewidth=2, color='r')
plt.show()
```

```python
df=2.72
mean,varian,skew,kurtosis=ss.t.stats(df,moments='mvsk')
x = np.linspace(ss.t.ppf(0.01, df),ss.t.ppf(0.99, df), 100)
plt.plot(x, ss.t.pdf(x, df),'b-', lw=5, alpha=0.6, label='t-student distribution')
plt.legend()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate some sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

# Create a 2x2 grid of subplots
fig, axs = plt.subplots(2, 2)

# Plot the data on each subplot
axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Sine')

axs[0, 1].plot(x, y2)
axs[0, 1].set_title('Cosine')

axs[1, 0].plot(x, y3)
axs[1, 0].set_title('Tangent')

axs[1, 1].remove()  # remove the last subplot

# Add a super title for the entire figure
fig.suptitle('Trigonometric Functions')

# Set the size of the figure
fig.set_size_inches(8, 6)

# Display the plot
plt.show()
```
