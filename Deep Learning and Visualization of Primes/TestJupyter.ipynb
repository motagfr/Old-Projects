```python
import math

def draw_circle_text(radius):
  for y in range(-radius, radius + 1):
    row = ""
    for x in range(-radius, radius + 1):
      if x*x + y*y <= radius*radius:
        row += "*"
      else:
        row += " "
    print(row)

draw_circle_text(5)
```
