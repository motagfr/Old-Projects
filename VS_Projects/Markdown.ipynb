<h1 style="color:blue; font-family:Nazanin;" dir="rtl"> Jupyter Notebook </h1>

<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
Jupyter Notebook یا Jupyter نوت‌بوک یک ابزار محیط توسعه یا IDE است که به طور خاص برای کار با زبان‌های برنامه‌نویسی مانند Python, R و Julia طراحی شده است (Jupyter مخفف حروف ابتدایی کلمات Julia, Python, R است). با استفاده از Jupyter Notebook می‌توانید کدهای خود را در قالب یک مستند تعاملی اجرا کنید و نتایج مربوطه را به همراه توضیحاتی که درباره کد ارائه می‌دهید، در یک محیط تکاملی و دوستانه بررسی کنید.
 </p>   
 
<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
Jupyter Notebook دارای مزایای زیادی است که شامل قابلیت اجرای تک خطی کد، دسترسی به کتابخانه‌های پرکاربرد، قابلیت نمایش تصاویر و ویدئوها، قابلیت ایجاد یادداشت‌های توضیحی و توضیحاتی در مورد کد، و قابلیت اشتراک‌گذاری نوت‌بوک‌ها با دیگران است. در کل، Jupyter Notebook به عنوان یک ابزار مفید برای برنامه‌نویسان، دانشجویان و محققان در زمینه‌های مختلف مورد استفاده قرار می‌گیرد.

<h1 style="color:blue; font-family:Nazanin;" dir="rtl"> اجزای یک نوت‌بوک Jupyter </h1>

<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
هر نوت‌بوک Jupyter از تعدادی سلول (cell) تشکیل شده است. هرسلول براساس محتوای خود می‌تواند بر دونوع باشد:
</p>   

<ol style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
<li>سلول کد (code): .این سلول حاوی کدهای برنامه (به‌صورت پیش‌فرض به زبان پایتون) است
</li>
<li>سلول مارک‌داون (Markdown): این سلول حاوی متن نوشته شده با مارک‌داون است.</li>
</ol>

<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
با فشردن همزمان دکمه‌های Shift + Enter محتوای سلول اجرا می‌شود.
اگر سلول حاوی کد برنامه باشد، خروجی این کد در زیر سلول نشان داده می‌شود. اگر سلول حاوی متن مارک‌داون باشد، متن پس از اعمال تنظیمات مارک‌داون نشان داده می‌شود. 
</p>

<h1 style="color:blue; font-family:Nazanin;" dir="rtl">مارک‌داون (Markdown) چیست؟</h1>

<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
Markdown یک زبان نشانه‌گذاری ساده است که برای نوشتن متن‌های ساده و قابل خواندن در وب و دیگر محیط‌ها طراحی شده است. این زبان نشانه‌گذاری توسط جان گروبر (John Gruber) و آرون شوایتز (Aaron Swartz) در سال 2004 ابداع شد.
 </p>   
 
<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
Markdown به کاربران اجازه می‌دهد تا متن خود را با استفاده از یک سری نشانه‌ها و علامت‌ها به صورت ساده و قابل خواندن تبدیل کنند. برای مثال، با استفاده از نشانه # می‌توانید عنوان‌های مختلف را ایجاد کنید و با استفاده از نشانه * و _ می‌توانید متن را تاکید کنید. همچنین، می‌توانید لیست‌ها، جداول، لینک‌ها، تصاویر و بسیاری از دیگر ویژگی‌های متنی را با استفاده از Markdown ایجاد کنید
</p>   

<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
Markdown به عنوان یک زبان نشانه‌گذاری ساده و قابل استفاده، در بسیاری از ابزارهای وب مانند وبلاگ‌ها، سیستم‌های مدیریت محتوا، نرم‌افزارهای نوشتاری و بسیاری دیگر از برنامه‌های کاربردی استفاده می‌شود.

</p>

<h1 style="color:blue; font-family:Nazanin;" dir="rtl"> درست‌کردن لیست</h1>

<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
برای درست کردن لیست‌های بدون ترتیب می‌توان از هریک از کاراکترهای *, + و یا - استفاده‌نمود:
</p>   

```markdown
* List item 1
* List item 2
* List item 3
```

* List item 1 using *
* List item 2 using *
* List item 3 using *


+ List item 1 using +
+ List item 2 using +
+ List item 3 using +


- List item 1 using -
- List item 2 using -
- List item 3 using -


<p style="font-family:Nazanin; font-size:120%; text-align:right;" dir="rtl">
برای ایجاد لیست های شماره دار و مرتب، تنها کافیست قبل از هرکدام از اعضای لیست یک عدد و نقطه قرار داد
1.
2.
...
    (می‌توان از یک عدد مشابه استفاده نمود):
 </p>  
 
